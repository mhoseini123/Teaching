# Fundamentals


## Task 1: Cloud Service Model Comparison Scenario

Imagine you are a solutions architect for a rapidly growing tech startup. The business needs to choose between IaaS, PaaS, and SaaS for three different projects: (1) Custom database infrastructure migration, (2) Rapid web app development with minimal infrastructure management, and (3) Enterprise email and collaboration tools. Write a short architectural brief (1–2 paragraphs per project) justifying the appropriate service delivery model for each scenario, detailing the balance of control versus management overhead.


## Task 2: Cloud Deployment Model Analysis

Examine your current organization or a well-known enterprise (e.g., a bank or a government healthcare portal). Determine whether a public, private, hybrid, or community cloud deployment model is best suited for their operations. Identify two primary security or regulatory constraints that influence this decision and explain how cloud characteristics (such as elasticity or measured service) apply.
